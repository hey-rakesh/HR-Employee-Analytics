```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"D:\Jupiter\Employee_project\data\emp_messy.csv", encoding="UTF-8")
```

```python
print(df.shape)
# print(df.duplicated().sum())
print(df.drop_duplicates(inplace=True))

```

```python
df.columns
df.dtypes
df.shape
```

*Employee ID*

```python
df["employee_id"] = df["employee_id"].str.strip().str.upper()
df["employee_id"] = df["employee_id"].str.replace("-","", regex=False).str.replace(" ", "", regex=False)
print(df["employee_id"].unique())
df["employee_id"].nunique()
# df["employee_id"].head(20)
```

*Name*

```python
df["name"].shape
df["name"] = df["name"].str.strip().str.lower().str.title()
print(df["name"].unique())
```

*Age*

```python
# df["age"].shape
print(df["age"].isnull().sum())
(df["age"].unique())
df["age"] = df["age"].replace({"unknown":np.nan,
                                   "?":np.nan,
                                   "Thirty":30,
                                   "30yrs":30,
                                   "25 years":25,
                                   "age 35":35,
                                   29.5:29})


df["age"] = pd.to_numeric(df["age"],errors="coerce")
print(df["age"].dtype)
df.loc[(df["age"] < 16) | (df["age"] > 70), "age"] = np.nan
print((df["age"].unique()))
df["age"] = df["age"].fillna(df["age"].median()).astype(int)
```

*gender*

```python

df["gender"] = df["gender"].replace({"Femal":"Female",
                                   "male":"Male",
                                   "?":np.nan,
                                   "Woman":"Female",
                                   "F":"Female",
                                   "female":"Female",
                                   "Man":"Male",
                                   "M":"Male"})
df["gender"] = df["gender"].replace(np.nan,"Unknown")

print(df["gender"].unique())
```

*department*

```python
df["department"].nunique()
# df["department"].unique()
df["department"] = df["department"].str.strip().replace({"it":"IT",
                          "I.T.":"IT",
                          "Marketting":"Marketing",
                          "SALES":"Sales",
                          "Research & Development":"R&D",
                          "Information Technology":"IT",
                          "Ops":"Operations",
                          "Finanace":"Finance",
                          "Finaace":"Finance",
                          "Human Resources":"HR",
                          "Cust Support":"Customer Support"})
df["department"].unique()
```

*job_role*

```python
df["job_role"].nunique()
df["job_role"].unique()
df["job_role"] = df["job_role"].str.strip().replace({"Analyst":"Data Analist",
                                    "software engineer":"Engineer",
                                    "Software Engg":"Engineer",
                                    "BDE": "Business Development Executive",
                                    "Sales Exec":"Sales Executive",
                                    "QA":"Quality Assurance",
                                    "HR Exec":"HR Executive",
                                    "HR Ececutive":"HR Executive"})
df["job_role"].unique()
```

*city*

```python
df["city"].nunique()
df["city"].unique()
df["city"] = df["city"].str.strip().replace({"Chennai,":"Chennai",
                                             "indore":"Indore",
                                             "Puna":"Pune",
                                             "Delhi":"Delhi NCR",
                                             "Hyderabad":"Hydrabad",
                                             "Ahmedabad":"Ahemdabad",
                                             "New Delhi":"Delhi NCR",
                                             "Bengaluru":"Bangalore",
                                             np.nan:"Unknown"})
df["city"].unique()
```

*education*

```python
df["education"].nunique()
df["education"].unique()
df["education"] = df["education"].str.strip().str.lower().str.title().replace({"BTECH":"B.Tech",
                                                                               "Mba":"MBA",
                                                                               "Mca":"MCA",
                                                                               "Bba":"BBA",
                                                                               "Phd":"PHD",
                                                                               "Be":"B.E.",
                                                                               "B Tech":"B.Tech",
                                                                               "Bcom":"B.Com",
                                                                               "M.B.A":"MBA",
                                                                               "Btech":"B.Tech",
                                                                               np.nan:"Unknown"})
df["education"].unique()
```

*experience_years*

```python
# df["experience_years"].nunique()
# df["experience_years"].unique()
print(df["experience_years"].isnull().sum())

df["experience_years"] = df["experience_years"].astype(str).str.strip().replace({"five":5,
                                                         "30 yrs":"30",
                                                         "3.5 yrs":"3.5",
                                                         "5 years":"5",
                                                         "10yrs":10,
                                                         "unknown":np.nan,
                                                         "?":np.nan,
                                                         "20+":20
})

df["experience_years"] = pd.to_numeric(df["experience_years"],errors="coerce")
df.loc[df["experience_years"] < 0, "experience_years"] = np.nan
df["experience_years"] = df["experience_years"].fillna(df["experience_years"].median())

print(df["experience_years"].unique())
print(df["experience_years"].isnull().sum())
```

*salary*

```python
# print(df["salary"].value_counts())
# df["salary"].isnull().sum()
# print(df["salary"].unique())
df["salary"] = df["salary"].replace({"unknown":np.nan,
                                     "?":np.nan,
                                     "N/A":np.nan
})

mask = df["salary"].astype("string").str.contains("crore", na=False)

df.loc[mask, "salary"] = (
    df.loc[mask, "salary"]
    .astype("string")
    .str.replace("crore", "", regex=False)
    .astype(float)
    * 10000000
)

mask = df["salary"].astype("string").str.contains("lpa|lakh|LPA", regex=True, na=False)

df.loc[mask, "salary"] = (
    df.loc[mask, "salary"]
    .astype("string")
    .str.replace("lpa", "", regex=False)
    .str.replace("LPA", "", regex=False)
    .str.replace("lakh", "", regex=False)
    .str.strip()
    .astype(float)
    * 100000
)
                            

mask = df["salary"].astype("string").str.endswith("k", na=False)
df.loc[mask, "salary"] = df.loc[mask, "salary"].astype("string").str.replace("k","", regex=False).astype(float) * 1000


df["salary"] = (
    df["salary"]
    .astype("string")
    .str.replace("₹", "", regex=False)
    .str.replace("rs.", "", regex=False)
    .str.replace("Rs. ", "", regex=False)
    .str.replace("inr", "", regex=False)
    .str.replace("INR", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)
df["salary"] = pd.to_numeric(df["salary"], errors="coerce")
df.loc[df["salary"] <= 0, "salary"] = np.nan
df["salary"] = df["salary"].fillna(df["salary"].median())
print(df["salary"].isnull().sum())
print(df["salary"].value_counts())
print(df["salary"].unique())
```
