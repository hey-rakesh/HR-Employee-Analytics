```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(r"D:\Jupiter\Employee_project\data\emp_messy.csv", encoding="UTF-8")
```

```python
print(df.shape)
# print(df.duplicated().sum())
print(df.drop_duplicates(inplace=True))

```

```python
df.columns
df.dtypes
df.shape
```

*Employee ID*

```python
df["employee_id"] = df["employee_id"].str.strip().str.upper()
df["employee_id"] = df["employee_id"].str.replace("-","", regex=False).str.replace(" ", "", regex=False)
print(df["employee_id"].unique())
df["employee_id"].nunique()
# df["employee_id"].head(20)
```

*Name*

```python
df["name"].shape
df["name"] = df["name"].str.strip().str.lower().str.title()
print(df["name"].unique())
```

*Age*

```python
# df["age"].shape
print(df["age"].isnull().sum())
(df["age"].unique())
df["age"] = df["age"].replace({"unknown":np.nan,
                                   "?":np.nan,
                                   "Thirty":30,
                                   "30yrs":30,
                                   "25 years":25,
                                   "age 35":35,
                                   29.5:29})


df["age"] = pd.to_numeric(df["age"],errors="coerce")
print(df["age"].dtype)
df.loc[(df["age"] < 16) | (df["age"] > 70), "age"] = np.nan
print((df["age"].unique()))
df["age"] = df["age"].fillna(df["age"].median()).astype(int)
```

*gender*

```python

df["gender"] = df["gender"].replace({"Femal":"Female",
                                   "male":"Male",
                                   "?":np.nan,
                                   "Woman":"Female",
                                   "F":"Female",
                                   "female":"Female",
                                   "Man":"Male",
                                   "M":"Male"})
df["gender"] = df["gender"].replace(np.nan,"Unknown")

print(df["gender"].unique())
```

*department*

```python
df["department"].nunique()
# df["department"].unique()
df["department"] = df["department"].str.strip().replace({"it":"IT",
                          "I.T.":"IT",
                          "Marketting":"Marketing",
                          "SALES":"Sales",
                          "Research & Development":"R&D",
                          "Information Technology":"IT",
                          "Ops":"Operations",
                          "Finanace":"Finance",
                          "Finaace":"Finance",
                          "Human Resources":"HR",
                          "Cust Support":"Customer Support"})
df["department"].unique()
```

*job_role*

```python
df["job_role"].nunique()
df["job_role"].unique()
df["job_role"] = df["job_role"].str.strip().replace({"Analyst":"Data Analist",
                                    "software engineer":"Engineer",
                                    "Software Engg":"Engineer",
                                    "BDE": "Business Development Executive",
                                    "Sales Exec":"Sales Executive",
                                    "QA":"Quality Assurance",
                                    "HR Exec":"HR Executive",
                                    "HR Ececutive":"HR Executive"})
df["job_role"].unique()
```

*city*

```python
df["city"].nunique()
df["city"].unique()
df["city"] = df["city"].str.strip().replace({"Chennai,":"Chennai",
                                             "indore":"Indore",
                                             "Puna":"Pune",
                                             "Delhi":"Delhi NCR",
                                             "Hyderabad":"Hydrabad",
                                             "Ahmedabad":"Ahemdabad",
                                             "New Delhi":"Delhi NCR",
                                             "Bengaluru":"Bangalore",
                                             np.nan:"Unknown"})
df["city"].unique()
```

*education*

```python
df["education"].nunique()
df["education"].unique()
df["education"] = df["education"].str.strip().str.lower().str.title().replace({"BTECH":"B.Tech",
                                                                               "Mba":"MBA",
                                                                               "Mca":"MCA",
                                                                               "Bba":"BBA",
                                                                               "Phd":"PHD",
                                                                               "Be":"B.E.",
                                                                               "B Tech":"B.Tech",
                                                                               "Bcom":"B.Com",
                                                                               "M.B.A":"MBA",
                                                                               "Btech":"B.Tech",
                                                                               np.nan:"Unknown"})
df["education"].unique()
```

*experience_years*

```python
# df["experience_years"].nunique()
# df["experience_years"].unique()
print(df["experience_years"].isnull().sum())

df["experience_years"] = df["experience_years"].astype(str).str.strip().replace({"five":5,
                                                         "30 yrs":"30",
                                                         "3.5 yrs":"3.5",
                                                         "5 years":"5",
                                                         "10yrs":10,
                                                         "unknown":np.nan,
                                                         "?":np.nan,
                                                         "20+":20
})

df["experience_years"] = pd.to_numeric(df["experience_years"],errors="coerce")
df.loc[df["experience_years"] < 0, "experience_years"] = np.nan
df["experience_years"] = df["experience_years"].fillna(df["experience_years"].median())

print(df["experience_years"].unique())
print(df["experience_years"].isnull().sum())
```

*salary*

```python
# print(df["salary"].value_counts())
# df["salary"].isnull().sum()
# print(df["salary"].unique())
df["salary"] = df["salary"].replace({"unknown":np.nan,
                                     "?":np.nan,
                                     "N/A":np.nan
})

mask = df["salary"].astype("string").str.contains("crore", na=False)

df.loc[mask, "salary"] = (
    df.loc[mask, "salary"]
    .astype("string")
    .str.replace("crore", "", regex=False)
    .astype(float)
    * 10000000
)

mask = df["salary"].astype("string").str.contains("lpa|lakh|LPA", regex=True, na=False)

df.loc[mask, "salary"] = (
    df.loc[mask, "salary"]
    .astype("string")
    .str.replace("lpa", "", regex=False)
    .str.replace("LPA", "", regex=False)
    .str.replace("lakh", "", regex=False)
    .str.strip()
    .astype(float)
    * 100000
)
                            

mask = df["salary"].astype("string").str.endswith("k", na=False)
df.loc[mask, "salary"] = df.loc[mask, "salary"].astype("string").str.replace("k","", regex=False).astype(float) * 1000


df["salary"] = (
    df["salary"]
    .astype("string")
    .str.replace("₹", "", regex=False)
    .str.replace("rs.", "", regex=False)
    .str.replace("Rs. ", "", regex=False)
    .str.replace("inr", "", regex=False)
    .str.replace("INR", "", regex=False)
    .str.replace(",", "", regex=False)
    .str.strip()
)
df["salary"] = pd.to_numeric(df["salary"], errors="coerce")
df.loc[df["salary"] <= 0, "salary"] = np.nan
df["salary"] = df["salary"].fillna(df["salary"].median())
print(df["salary"].isnull().sum())
print(df["salary"].value_counts())
print(df["salary"].unique())
```

*performance rating*

```python
print(df["performance_rating"].unique())
# print(df["performance_rating"].value_counts())
df["performance_rating"] = df["performance_rating"].replace({"three":3,
                                                             "Good":4,
                                                             "Average":6,
                                                             "Poor":2,
                                                             "4 stars":4,
                                                             "Excellent":9,
                                                             "Rating 4":4,
                                                             "5/5":5
})

df["performance_rating"] = pd.to_numeric(df["performance_rating"], errors="coerce")
df["performance_rating"] = df["performance_rating"].fillna(df["performance_rating"].median())
print(df["performance_rating"].unique())
```

*overtime*

```python
print(df["overtime"].unique())
df["overtime"] = df["overtime"].str.strip().replace({"Y":"Yes",
                                        "N":"No",
                                        "False":"No",
                                        "TRUE":"Yes",
                                        "yes":"Yes",
                                        "YES":"Yes"
})
print(df["overtime"].unique())
```

*work_mode*

```python
print(df["work_mode"].unique())
df["work_mode"] = df["work_mode"].str.strip().replace({"Work From Home":"WHF",
                                                       "remote":"Remote",
                                                       "REMOTE":"Remote",
                                                       "On-site":"Onsite",
                                                       "Home":"WFH",
                                                       "WHF":"WFH"
})
print(df["work_mode"].unique())
```

*projects_handled*

```python
print(df["projects_handled"].unique())
df["projects_handled"] = df["projects_handled"].replace({"zero":0,
                                                        "none":0,
                                                        "20 projects":20,
                                                        "5-6":5.5,
                                                        "3.0":3,
                                                        "15+":15,
                                                        "5 projects":5,
                                                        "-1":np.nan,
                                                        "many":np.nan,
                                                        "unknown":np.nan})
df["projects_handled"] = pd.to_numeric(df["projects_handled"], errors="coerce")
df["projects_handled"] = df["projects_handled"].fillna(df["projects_handled"].median())
print(df["projects_handled"].unique())
```

*training_hours*

```python
# print(df["training_hours"].unique())
df["training_hours"] = df["training_hours"].replace({"none":0,
                                                     "1 week":40,
                                                     "5 days":40,
                                                     "20h":20,
                                                     "40 hrs":40,
                                                     "zero":0,
                                                     "40-50":45,
                                                     "unknown":np.nan,
                                                     "200+":200,
                                                     "100 hours":100,
                                                     })

df["training_hours"] = df["training_hours"].fillna(df["training_hours"].median())
# print(df["training_hours"].unique())
# print(df["training_hours"].isnull().sum())
```

*promotion_status*

```python
print(df["promotion_status"].unique())
df["promotion_status"] = df["promotion_status"].str.strip().replace({"yes":"Yes",
                                                                     "Not Promoted":"No",
                                                                     "Y":"Yes",
                                                                     "YES":"Yes",
                                                                     "unknown":np.nan,
                                                                     "N":"No",
                                                                     "Promoted":"Yes",
                                                                     "Pending":"No"})
df["promotion_status"] = df["promotion_status"].fillna(df["promotion_status"].mode()[0])
print(df["promotion_status"].unique())
df["promotion_status"].value_counts(dropna=False)
```

*attrition_status*

```python
print(df["attrition_status"].unique())
df["attrition_status"] = df["attrition_status"].str.strip().replace({"Active":"No",
                                                                    "Resigned":"Yes",
                                                                    "Y":"Yes",
                                                                    "yes":"Yes",
                                                                    "Stayed":"No",
                                                                    "Left":"Yes",
                                                                    "unknown":np.nan,
                                                                    "N":"No",
                                                                    "YES":"Yes"})

df["attrition_status"] = df["attrition_status"].fillna(df["attrition_status"].mode()[0])
print(df["attrition_status"].unique())
```

*joining_date*

```python
# print(df["joining_date"].unique())
df["joining_date"] = df["joining_date"].replace({"?":np.nan,
                                                 "unknown":np.nan,
                                                 "4":np.nan})

df["joining_date"] = pd.to_datetime(df["joining_date"], errors="coerce", dayfirst=True, format="mixed")
print(df["joining_date"].unique())
print(df["joining_date"].isnull().sum())
```

```python
# ==================================================
# FINAL DATASET CHECKLIST
# ==================================================


# 1. Dataset Shape
print("SHAPE:")
print(df.shape)


# 2. Column Names
print("\nCOLUMN NAMES:")
print(df.columns.tolist())


# 3. Data Types
print("\nDATA TYPES:")
print(df.dtypes)


# 4. Dataset Information
print("\nDATASET INFO:")
df.info()


# 5. Missing Values
print("\nMISSING VALUES:")
print(df.isnull().sum())


# 6. Total Missing Values
print("\nTOTAL MISSING VALUES:")
print(df.isnull().sum().sum())


# 7. Duplicate Rows
print("\nDUPLICATE ROWS:")
print(df.duplicated().sum())


# 8. Unique Values Count
print("\nUNIQUE VALUES COUNT:")
print(df.nunique())


# 9. Numeric Columns Summary
print("\nNUMERIC COLUMNS SUMMARY:")
print(df.describe())


# 10. Categorical Columns Summary
print("\nCATEGORICAL COLUMNS SUMMARY:")
print(df.describe(include="object"))


# ==================================================
# 11. CHECK FINAL CATEGORICAL VALUES
# ==================================================

categorical_columns = [
    "gender",
    "department",
    "city",
    "education_level",
    "job_role",
    "promotion_status",
    "attrition_status"
]

print("\nCATEGORICAL COLUMN VALUES:")

for col in categorical_columns:
    print(f"\n{col.upper()}:")
    print(df[col].unique())
    print("Missing:", df[col].isnull().sum())


# ==================================================
# 12. CHECK NUMERIC RANGE
# ==================================================

numeric_columns = [
    "age",
    "experience_years",
    "salary",
    "training_hours",
    "backlogs"
]

print("\nNUMERIC COLUMN MIN/MAX:")

for col in numeric_columns:
    print(f"\n{col.upper()}")
    print("Min:", df[col].min())
    print("Max:", df[col].max())


# ==================================================
# 13. CHECK NEGATIVE VALUES
# ==================================================

print("\nNEGATIVE VALUE CHECK:")

for col in [
    "age",
    "experience_years",
    "salary",
    "training_hours",
    "backlogs"
]:
    print(
        f"{col} negative values:",
        (df[col] < 0).sum()
    )


# ==================================================
# 14. CHECK AGE RANGE
# ==================================================

print("\nAGE RANGE CHECK:")

print(
    "Invalid Age Values:",
    ((df["age"] < 18) | (df["age"] > 100)).sum()
)


# ==================================================
# 15. CHECK EXPERIENCE RANGE
# ==================================================

print("\nEXPERIENCE CHECK:")

print(
    "Invalid Experience:",
    (df["experience_years"] < 0).sum()
)


# ==================================================
# 16. CHECK DATE COLUMN
# ==================================================

print("\nDATE COLUMN CHECK:")

date_columns = [
    "joining_date"
]

for col in date_columns:
    print(f"\n{col}:")
    print("Missing:", df[col].isnull().sum())
    print("Data type:", df[col].dtype)
    print("Minimum Date:", df[col].min())
    print("Maximum Date:", df[col].max())


# ==================================================
# 17. CHECK DUPLICATE EMPLOYEE IDs
# ==================================================

print("\nDUPLICATE EMPLOYEE ID CHECK:")

print(
    df["employee_id"].duplicated().sum()
)


# ==================================================
# 18. FINAL RANDOM SAMPLE
# ==================================================

print("\nFINAL DATA SAMPLE:")

print(df.sample(10, random_state=42))
```
