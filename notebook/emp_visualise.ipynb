# HR Employee dataset visualisation

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"D:\Jupiter\Employee_project\data\emp_clean.csv", encoding="UTF-8")
```

```python
plt.figure(figsize=(10, 6))

ax = sns.countplot(data=df, x="department")

plt.bar_label(
    ax.containers[0],
    fmt="%.0f",
    padding=3
)
plt.xticks(rotation=90)
plt.show()

```

```python
dept_high_salary = df.groupby("department")["salary"].mean()
plt.figure(figsize=(10, 6))
ax = sns.barplot(x=dept_high_salary.index, y=dept_high_salary.values)
plt.xticks(rotation=90)
plt.bar_label(
    ax.containers[0],
    fmt="%.0f",
    padding=3

)
plt.show()
```

```python
sns.histplot(data=df, x= "salary", bins=8, kde=True)

plt.show()
```

```python
plt.figure(figsize=(10, 4))
sns.boxplot(data=df, x="department", y="salary")
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(data=df, x="experience_years", y="salary")
plt.show()
```

```python
df["joining_date"] = pd.to_datetime(df["joining_date"])
emp_per_year = df["joining_date"].dt.year.value_counts()
sns.lineplot(x=emp_per_year.index, y=emp_per_year.values, marker="o")
plt.show()
```

```python
plt.figure(figsize=(10,8))
ax = sns.countplot(data=df, x="department", hue="gender")
for container in ax.containers:
    ax.bar_label(
        container,
        fmt="%.0f",
        padding=3
    )

plt.xticks(rotation=90)
plt.show()
```

```python

attrn_count = df["attrition_status"].value_counts()
plt.figure(figsize=(8,6))
plt.pie(attrn_count.values,
        labels=attrn_count.index,
        autopct="%1.1f%%",
        startangle=90)
plt.show()
```

```python
plt.figure(figsize=(10, 8))

correlation = df[["age", "experience_years", "salary"]].corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.show()
```

```python
plt.figure(figsize=(10,8))
sns.barplot(data=df, x="department", y="salary", hue="gender", errorbar=None)
plt.xticks(rotation=90)
plt.show()
```

```python
sns.violinplot(data=df, x="attrition_status", y="salary")
plt.show()
```

```python
attrn_dept = df[df["attrition_status"] == "Yes"].groupby("department").size()
total_emp_dept = df.groupby("department")["employee_id"].size()
attrn_rate = ((attrn_dept / total_emp_dept) * 100).round(2)
plt.figure(figsize=(10,6))
ax = sns.barplot(x=attrn_rate.index, y=attrn_rate.values)
plt.bar_label(
    ax.containers[0],
    fmt="%.2f",
    padding=3
)
plt.xticks(rotation=90)
plt.ylabel("Attrition Rate (%)")
plt.show()
```

```python
plt.figure(figsize=(10,8))
sns.boxplot(data=df, x="job_role", y="salary")
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(data=df, x="attrition_status", y="salary")
plt.show()
```

```python
df["age_group"] = pd.cut(
    df["age"],
    bins=[18,25,35,45,55,65],
    labels=["18-25", "26-35", "36-45", "46-55", "56-65"]
)
plt.figure(figsize=(8, 6))
ax = sns.countplot(data=df, x="age_group")
plt.bar_label(
    ax.containers[0],
    fmt="%.0f",
    padding=3
)
plt.show()
```

```python
plt.figure(figsize=(8,6))
attred_emp = df[df["attrition_status"] == "Yes"]
ax = sns.countplot(data=attred_emp, x="job_role")
plt.xticks(rotation=90)
plt.bar_label(
    ax.containers[0],
    fmt="%.0f",
    padding=3
)
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.barplot(data=df, x="department",y="performance_rating", errorbar=None)
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(10,8))
sns.scatterplot(data=df, x="age", y="performance_rating", hue="gender")
plt.show()
```

```python
avg_sal_exp = df.groupby("experience_years")["salary"].mean().reset_index()
plt.figure(figsize=(10,8))
sns.lineplot(data=avg_sal_exp, x="experience_years", y="salary", marker="o")
plt.show()
```

```python
plt.figure(figsize=(10,8))
sns.histplot(data=df, x="salary", hue="gender", multiple="layer", kde=True)
plt.show()
```

```python
plt.figure(figsize=(10,8))
sns.scatterplot(data=df, x="experience_years", y="salary", hue="department")
plt.show()
```

```python
plt.figure(figsize=(8,6))
attred = df["attrition_status"].value_counts()
plt.pie(attred.values,labels=attred.index,autopct="%1.1f%%", startangle=90)
plt.show()
```

```python
plt.figure(figsize=(10,8))
correlation = df.corr(numeric_only=True)
sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(data=df, x= "department",y="salary", hue="attrition_status",errorbar=None)
plt.xticks(rotation=90)
plt.show()
```

```python
df["joining_year"] = pd.to_datetime(df["joining_date"]).dt.year
group = (df.groupby(["joining_year", "department"]).size().reset_index(name="employee_count"))
plt.figure(figsize=(10, 8))

sns.lineplot(
    data=group,
    x="joining_year",
    y="employee_count",
    hue="department",
    marker="o"
)

plt.show()
```
